1. Program to write explain different types of Inheritances using classes.

```python
#single inheritance
class animal:
    def walks(self):
        return 'animal can walk'
class dog(animal):
    pass
obj=dog()
obj.walks()
```

```python
#multilevel inheritance
class man:
    def run(self):
        print('man sneaks')
class boy(man):
    def walk(self):
        print ('boy walks')
class baby(boy):
    def crawl(self):
        print ('baby crawls')
obj=baby()
obj.crawl()
obj.run()
obj.walk()
```

```python
#multipe inheritance
class father:
    def vehicle(self):
        print('car')
class mother:
    def jewels(self):
        print('gold')
class child(father,mother):
    pass
obj=child()
obj.jewels()
obj.vehicle()
```

```python
class father:
    def property(self):
        print('Father Property')
class mother:
    def property(self):
        print('Mother Property')
class child(father,mother):
    pass
obj=child()
obj.property()
```

```python
class father:
    def property(self):
        print('Father Property')
class mother:
    def property(self):
        print('Mother Property')
class child(father,mother):
    pass
obj=child()
father.property(obj)
mother.property(obj)
```

```python
#hierarchical inheritance
class parent:
    def property(self):
        print( 'land')
class child1(parent):
    pass
class child2(parent):
    pass
obj=child1()
obj1=child2()
obj.property()
obj1.property()
```

```python
#hybrid inheritance
class A:
    def showA(self):
        print("class A")
class B(A):
    def showB(self):
        print("class B")
class C(A):
    def showC(self):
        print("class C")
class D(B, C):
    def showD(self):
        print("class D")
obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

12. Create a BankAccount class with attributes for account number, owner name, and balance. Include methods for deposit, withdraw, and check balance

```python
class BankAccount:
    def __init__(self, acc_no, owner, balance):
        self.acc_no = acc_no
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(f'{amount} deposited')
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f'{amount} withdrawn')
        else:
            print("Insufficient Balance")
    def checkBalance(self):
        print("Balance:", self.balance)
obj = BankAccount(101, "Sandeep", 5000)
obj.deposit(2000)
obj.withdraw(3000)
obj.checkBalance()
```

13. Create a Library class to manage books. Implement methods to add books, borrow books, and return books.

```python
class Library:
    def __init__(self,books):
        self.books =books
    def addBook(self, book):
        self.books.append(book)
        print(f'{book} added')
    def borrowBook(self, book):
        if book in self.books:
            self.books.remove(book)
            print(f'{book} borrowed')
        else:
            print("Book not available")
    def returnBook(self, book):
        self.books.append(book)
        print(f'{book} returned')
    def showBooks(self):
        print("Books:", self.books)
obj = Library(['story','non-fictional'])
obj.addBook("python")
obj.addBook("java")
obj.borrowBook("python")
obj.returnBook("python")
obj.showBooks()
```

15. Design a Shape hierarchy with a base class and derived classes for Circle, Rectangle, and Triangle. Include methods to calculate area and perimeter.

```python
class Shape:
    def area(self):
        pass
    def perimeter(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    def perimeter(self):
        return 2 * 3.14 * self.radius
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
class Triangle(Shape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):
        return self.a + self.b + self.c
c = Circle(5)
r = Rectangle(4, 6)
t = Triangle(3, 4, 5)
print(c.area())
print(c.perimeter())
print(r.area())
print(r.perimeter())
print(t.perimeter())
```

20. Design an abstract MediaPlayer class with methods like play(), pause(), and stop(). Implement concrete classes for different media types.

```python
from abc import ABC, abstractmethod
class MediaPlayer(ABC):
    @abstractmethod
    def play(self):
        pass
    @abstractmethod
    def pause(self):
        pass
    @abstractmethod
    def stop(self):
        pass
class AudioPlayer(MediaPlayer):
    def play(self):
        print("Audio Playing")
    def pause(self):
        print("Audio Paused")
    def stop(self):
        print("Audio Stopped")
class VideoPlayer(MediaPlayer):
    def play(self):
        print("Video Playing")
    def pause(self):
        print("Video Paused")
    def stop(self):
        print("Video Stopped")
a = AudioPlayer()
v = VideoPlayer()
a.play()
a.pause()
v.play()
v.stop()
```
