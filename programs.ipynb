14. Program to Find Sum of Digits of a Number using Recursion.

```python
def sum_digits(number):

    if number == 0:
        return 0

    return number % 10 + sum_digits(number // 10)


number = int(input("Enter number:"))

print("Sum of digits =", sum_digits(number))
```

15. Program to print the sum of alternative digits in a given number?

```python
number = int(input("Enter a number: "))

alternative_sum = 0

take_digit = True

while number > 0:
    digit = number % 10
    if take_digit:
        alternative_sum += digit
    take_digit = not take_digit
    number = number // 10
print("alternate digit sum=", alternative_sum)
```

19. Program to add even and odd digits of a given number?

```python
number=int(input('enter number'))
even_sum=0
odd_sum=0
while number>0:
    current_digit=number%10
    if current_digit%2==0:
        even_sum+=current_digit
    else:
        odd_sum+=current_digit
    number=number//10
print('odd sum',odd_sum)
print('even sum',even_sum)
```

22. Program to print n prime numbers.

```python
n=int(input("enter number "))
count=0
number=2
while count<n:
    is_prime = True
    for divisor in range(2, number):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        print(number)
        count += 1
    number += 1
```

23. Program to print perfect numbers between m to n.

```python
m = int(input("Enter starting number: "))
n = int(input("Enter ending number: "))
for number in range(m, n + 1):
    factor_sum = 0
    for divisor in range(1, number):
        if number % divisor == 0:
            factor_sum += divisor
    if factor_sum == number:
        print(number)
```

24. Program to print n perfect numbers

```python
n = int(input("Enter value of n: "))
count = 0
number = 1
while count < n:
    factor_sum = 0
    for divisor in range(1, number):
        if number % divisor == 0:
            factor_sum += divisor
    if factor_sum == number:
        print(number)
        count += 1
    number += 1
```

 25. Program to Print Armstrong Number between 1 to 1000.

```python
for number in range(1, 1001):
    original_number = number
    digit_cube_sum = 0
    while number > 0:
        digit = number % 10
        digit_cube_sum += digit ** 3
        number = number // 10
    if digit_cube_sum == original_number:
        print(original_number)
    number = original_number
```
