14. Program to Find Sum of Digits of a Number using Recursion.

```python
def sum_digits(number):

    if number == 0:
        return 0

    return number % 10 + sum_digits(number // 10)


number = int(input("Enter number:"))

print("Sum of digits =", sum_digits(number))
```

15. Program to print the sum of alternative digits in a given number?

```python
number = int(input("Enter a number: "))

alternative_sum = 0

take_digit = True

while number > 0:
    digit = number % 10
    if take_digit:
        alternative_sum += digit
    take_digit = not take_digit
    number = number // 10
print("alternate digit sum=", alternative_sum)
```

28. Class to Read File and Count Vowels, Consonants, Words, Numbers, Special Characters

```python
class FileAnalyzer:
    def __init__(self, filename):
        self.filename = filename

    def analyze(self):
        file = open(self.filename, "r")
        data = file.read()
        file.close()

        vowels = 0
        consonants = 0
        numbers = 0
        special = 0

        for ch in data:
            if ch.lower() in "aeiou":
                vowels += 1
            elif ch.isalpha():
                consonants += 1
            elif ch.isdigit():
                numbers += 1
            elif not ch.isspace():
                special += 1

        words = len(data.split())

        print("File Data:", data)
        print("Vowels:", vowels)
        print("Consonants:", consonants)
        print("Words:", words)
        print("Numbers:", numbers)
        print("Special Characters:", special)
```

29. Student Class using Dictionary and Search by Gmail

```python
class Student:
    def __init__(self):
        self.students = {}

    def add_student(self, student_id, name, email, phone, age):
        self.students[email] = {
            "ID": student_id,
            "Name": name,
            "Phone": phone,
            "Age": age
        }

    def search_by_email(self, email):
        if email in self.students:
            print("Student Found:")
            print("Email:", email)
            print("Details:", self.students[email])
        else:
            print("Student not found")


# Driver Code
s = Student()

s.add_student(101, "Sandeep", "sandeep@gmail.com", "9876543210", 21)
s.add_student(102, "Rahul", "rahul@gmail.com", "9123456780", 22)

email = input("Enter gmail to search: ")
s.search_by_email(email)
```

```python
import pymysql


# ================= CUSTOM EXCEPTIONS =================

class InvalidAccountNumber(Exception):
    pass


class InvalidPassword(Exception):
    pass


class DepositLimitException(Exception):
    pass


class InsufficientBalanceException(Exception):
    pass


class ReceiverAccountNotFound(Exception):
    pass


class SameAccountTransfer(Exception):
    pass


class TransactionLimitException(Exception):
    pass


# ================= DATABASE CONNECTION =================

conn = pymysql.connect(
    host='localhost',
    user='root',
    password='blazext_7_7',
    database='bank'
)

cursor = conn.cursor()


# ================= HELPER FUNCTIONS =================

def get_balance(acc_no):

    cursor.execute(
        '''
        select balance from customer
        where account_number=%s
        ''',
        (acc_no,)
    )

    return cursor.fetchone()[0]


def account_exists(acc_no):

    cursor.execute(
        '''
        select * from customer
        where account_number=%s
        ''',
        (acc_no,)
    )

    return cursor.fetchone()


# ================= LOGIN FUNCTION =================

def login():

    attempts = 0

    while attempts < 3:

        try:

            acc_no = int(
                input("Enter Account Number: ")
            )

            password = input(
                "Enter Password: "
            )

            account = account_exists(acc_no)

            if not account:

                raise InvalidAccountNumber(
                    "Invalid Account Number"
                )

            db_password = account[4]

            if password != db_password:

                raise InvalidPassword(
                    "Invalid Password"
                )

            print("\nWELCOME TO AXIS BANK")

            return acc_no

        except (
            InvalidAccountNumber,
            InvalidPassword
        ) as e:

            attempts += 1

            print(e)

            print(
                "Attempts Left:",
                3 - attempts
            )

    print(
        "\nAccount Blocked!"
    )

    print(
        "Try again after 1 hour."
    )

    return None


# ================= DEPOSIT FUNCTION =================

def deposit(acc_no):

    try:

        amount = float(
            input("Enter Deposit Amount: ")
        )

        if amount < 500:

            raise DepositLimitException(
                "Minimum deposit amount is 500"
            )

        cursor.execute(
            '''
            update customer
            set balance=balance+%s
            where account_number=%s
            ''',
            (amount, acc_no)
        )

        conn.commit()

        print(
            "Amount Deposited Successfully"
        )

        print(
            "Available Balance:",
            get_balance(acc_no)
        )

    except DepositLimitException as e:

        print(e)


# ================= WITHDRAW FUNCTION =================

def withdraw(acc_no):

    try:

        amount = float(
            input("Enter Withdraw Amount: ")
        )

        balance = get_balance(acc_no)

        if amount > balance:

            raise InsufficientBalanceException(
                "Insufficient Balance"
            )

        cursor.execute(
            '''
            update customer
            set balance=balance-%s
            where account_number=%s
            ''',
            (amount, acc_no)
        )

        conn.commit()

        print(
            "Withdraw Successful"
        )

        print(
            "Remaining Balance:",
            get_balance(acc_no)
        )

    except InsufficientBalanceException as e:

        print(e)


# ================= CHECK BALANCE FUNCTION =================

def check_balance(acc_no):

    print(
        "Available Balance:",
        get_balance(acc_no)
    )


# ================= TRANSFER FUNCTION =================

def transfer_money(acc_no):

    try:

        receiver_acc = int(
            input(
                "Enter Receiver Account Number: "
            )
        )

        if receiver_acc == acc_no:

            raise SameAccountTransfer(
                "Cannot transfer to same account"
            )

        receiver = account_exists(receiver_acc)

        if not receiver:

            raise ReceiverAccountNotFound(
                "Receiver Account Not Found"
            )

        amount = float(
            input(
                "Enter Transfer Amount: "
            )
        )

        if amount < 500:

            raise TransactionLimitException(
                "Minimum Transaction amount is 500"
            )

        sender_balance = get_balance(acc_no)

        if amount > sender_balance:

            raise InsufficientBalanceException(
                "Insufficient Balance"
            )

        # deduct sender balance

        cursor.execute(
            '''
            update customer
            set balance=balance-%s
            where account_number=%s
            ''',
            (amount, acc_no)
        )

        # add receiver balance

        cursor.execute(
            '''
            update customer
            set balance=balance+%s
            where account_number=%s
            ''',
            (amount, receiver_acc)
        )

        conn.commit()

        print(
            "Transaction Successful"
        )

        print(
            "Remaining Balance:",
            get_balance(acc_no)
        )

    except (
        ReceiverAccountNotFound,
        InsufficientBalanceException,
        SameAccountTransfer,
        TransactionLimitException
    ) as e:

        conn.rollback()

        print(e)


# ================= MAIN MENU FUNCTION =================

def main_menu(acc_no):

    while True:

        print("\nMENU")

        print("1. Deposit")

        print("2. Withdraw")

        print("3. Check Balance")

        print("4. Transfer Money")

        print("5. Exit")

        choice = int(
            input("Enter Choice: ")
        )

        if choice == 1:

            deposit(acc_no)

        elif choice == 2:

            withdraw(acc_no)

        elif choice == 3:

            check_balance(acc_no)

        elif choice == 4:

            transfer_money(acc_no)

        elif choice == 5:

            print(
                "Thank You For Banking With Us"
            )

            break

        else:

            print(
                "Invalid Option"
            )


# ================= DRIVER CODE =================

acc_no = login()

if acc_no:

    main_menu(acc_no)


# ================= CLOSE CONNECTION =================

cursor.close()

conn.close()
```
