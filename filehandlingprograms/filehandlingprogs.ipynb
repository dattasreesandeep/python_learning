18. Program to count vowels, consonants, digits, and special characters present in a text file separately.


```python
with open("data.txt", "r") as file:
    content = file.read()
vowels = 0
consonants = 0
digits = 0
special = 0
for ch in content:
    if ch.lower() in 'aeiou':
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Special Characters:", special)
```

25. Program to identify and display the most repeated word present in a text file with its
frequency.


```python
with open("data.txt","r") as file:
    words=file.read().split()
max_word=''
max_count=0
for word in words:
    count=words.count(word)
    if count>max_count:
        max_count=count
        max_word=word
print("Most Repeated Word:",max_word)
print("Frequency:",max_count)
```

```python
with open("data.txt", "r") as file:

    words = file.read().split()


frequency = {}


for word in words:

    frequency[word] = frequency.get(word, 0) + 1


most_repeated = max(frequency, key=frequency.get)


print("Most Repeated Word:", most_repeated)

print("Frequency:", frequency[most_repeated])
```

28. Program to identify repeated lines in a file and display the duplicated lines only once.


```python
with open("data.txt", "r") as file:
    lines = file.readlines()
seen = set()
duplicates = set()
for line in lines:
    line = line.strip()
    if line in seen:
        duplicates.add(line)
    else:
        seen.add(line)
print("Duplicate Lines:")
for line in duplicates:
    print(line)
```

30. Program to find and display all possible anagram words available in a text file.


```python
with open("data.txt","r") as file:
    words=file.read().split()

anagrams={}

for word in words:

    clean=''

    for ch in word:
        if ch.isalpha():
            clean=clean+ch

    if clean=='':
        continue

    key=''.join(sorted(clean.lower()))

    anagrams.setdefault(key,[]).append(word)

print("Anagram Groups:")

for value in anagrams.values():
    if len(value)>1:
        print(value)
```

33. Program to identify and display the top 10 most frequently used words present in a text file.

```python
with open("data.txt", "r") as file:
    words = file.read().split()
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1
sorted_words = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)
print("Top 10 Frequent Words:")
for word, count in sorted_words[:10]:
    print(word, ":", count)
```
