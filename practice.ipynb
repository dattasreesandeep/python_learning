```python
for i in range(2,51):
    is_prime=True
    for j in range(2,i):
        if i%j==0:
            is_prime=False
            break
    if is_prime:
        print(i)
```

```python
n=int(input("enter number"))
count=0
num=2
sum=0
while count<n:
    is_prime=True
    for i in range(2,num):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        print(num)
        sum+=num
        count+=1
    num+=1
print(sum)
```
