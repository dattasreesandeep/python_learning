1. Write a Python program to find all three, four, and five-character words in a string.


```python
import re
text='hi my name is sandeep and i have been learning python for 3 weeks'
result=re.findall(r'\b\w{3,5}\b',text)
print(result)
```

2. Write a Python script to transform a camelCase identifier into snake_case and then print
the result.


```python
import re
text = "camelCaseIdentifier"
result=re.sub(r'([a-z])([A-Z])',r'\1_\2',text).lower()
print(result)
```

3. Write a Python program to find URLs in a string.


```python
import re
text='this are my profiles www.google.com and https://github.com'
result=re.findall(r'https?://\S+|www.\S+',text)
print(result)
```

4. Write a Python program to remove everything except alphanumeric characters from a
string.


```python
import re
text='this are my profiles www.google.com and https://github.com sandeep*%*'
result=re.sub(r'[^a-zA-Z0-9]','',text)
print(result)
```

5. Write a Python program to split a string with multiple delimiters.


```python
import re
text = "Python,Java;C++ JavaScript.sql"
result=re.split(r'[,; .]+',text)
print(result)
```

6. Write a Python program to replace whitespaces with an underscore and vice versa.


```python
import re
text = "Python Java C_JavaScript_sql "
text=re.sub(r' +','_',text)
print(text)
```

```python
# now viceversa at a time
import re
text = "Python Java C_JavaScript_sql "
text=re.sub(r' +|_+',lambda x: '_' if x.group().strip() =='' else ' ',text)
print(text)
```

7. Write a Python program that matches a string that has an a followed by zero or more b's.


```python
#a string must match the only it is returned
import re

text = "abbb"

result = re.fullmatch(r'ab*', text)

if result:
    print("Matched")
else:
    print("Not Matched")
```

```python
#finds all patterns tht has a having zero b or b's
import re
text='a0 abbbbbb ffefefe bbbb a a00 ab abb'
text=re.findall(r'\bab*\b',text)
print(text)
```

8. Write a Python program that matches a string that has an a followed by zero or one 'b'.

9. Write a Python program that matches a string with an a followed by two to three 'b'.


```python
import re 
text='abbb'
text=re.fullmatch(r'ab{2,3}',text)
if text:
    print ('match')
else:
    print('notamatch')
```

13. Write a Python program to get the year, month, and date from a URL.


```python
import re
text = "https://example.com/2025/05/20/article"
t=re.search(r'(\d{4})/(\d{2})/(\d{2})',text)
print(t.group())
print(t.group(1))
print(t.group(2))
print(t.group(3))
```

```python
import re
text = "https://example.com/2025/05/20/article"
t=re.findall(r'(\d{4})/(\d{2})/(\d{2})',text)
print(t)
print(t[0][0])
print(t[0][1])
print(t[0][2])
```

14. Write a Python program to separate and print the numbers and their position in a given
string.


```python
import re
text = "Python 123 Java 456"
for match in re.finditer(r'(\d+)', text):
    print(
        "Number:", match.group(),
        "Position:", match.start()
    )
```
