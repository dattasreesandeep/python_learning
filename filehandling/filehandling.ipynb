```python
#opening and using read operation and closing file
file= open('data.txt','r')
content=file.read()
print(content)
file.close()
```

```python
#opening and performing read operation in a professional way it doesnt require closing and prvents memory leaks and resource locking issues
with open ('data.txt','r') as file:
    print(file.read())
```

```python
#reading line by line
with open ('data.txt','r') as file:
    for line in file:
        print(line.strip()) #line by line prints new line tooo so .strip() allows to removes new lines between lines
```

```python
#replacing old text with new text
with open ('data.txt','w') as file:
    file.write('hello sandeep\n')
    file.write('i am jello\n')
    file.write('i like to hangout with my fellow.')
```

```python
#better way with out deleting old text with appending new text
with open ('data.txt','a') as file:
    file.write('\nappend the new fellow too')
```

```python
#writing list of lines
lines=['first line\n','second line\n']
with open ('data.txt','a') as file:
    file.writelines(lines)
```

```python
#reading and writing binary
data=b'1,2,3,4'
with open ('data.bin','wb') as file:
    file.write(b'hello')
    file.write(b'\n'+data)
```

```python
#python objects to binary
import pickle
data=[1,2,3,4,5]
with open ('data.bin','wb') as file:
    pickle.dump(data,file)
```

```python
#reading pickled object
import pickle
with open('data.bin','rb') as file:
    print(pickle.load(file))
```

```python
#writing to binary file
data = b'\x00\x01\x02\x03\x04' #This code is written in raw binary bytes into binary file
with open('example.bin','wb')as file:
    file.write(data)
```

```python
#reading binary 
with open('example.bin','rb')as file:
    print(file.read())
```

```python
#read data from one file and write it into another file
with open('source.txt','r')as source_file:
    content=source_file.read()
with open('destination.txt','w')as destination_file:
    destination_file.write(content)
```

```python
#read text file and count number of lines, words and character in a file and return them
def file_count(file_name):
    with open (file_name,'r') as file:
        print('this is the content of file:\n'+file.read())
        file.seek(0)
        lines=file.readlines()
        lines_count=len(lines)
        words_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line.replace(' ','')) for line in lines)
        return lines_count,words_count,char_count
lines1,words,characters=file_count('file1.txt')
print(f'lines:{lines1},words: {words},characters: {characters}')
```
