1. Write a program to handle exceptions with Try-except Block.

```python
try:
    a=int(input('enter first number'))
    b=int(input('enter second number'))
    num=a/b
    print(num)
except ZeroDivisionError:
    print('enter number greater than zero')
except ValueError:
    print('enter a number')
```

2. Write a program to handle exceptions with Multiple except Block.

```python
try:
    num1 =int(input("first Number"))
    num2 = int(input("second Number"))
    result= num1/num2
    numbers= [10,20, 30]
    print(numbers[2])
    print("result=",result)
except ZeroDivisionError:
    print("cannot divide by zero")
except ValueError:
    print("enter number only")
except IndexError:
    print("list index out of range")
```

3. Write a program to handle exceptions with Nested Try.

```python
try:
    print("Outer Try Block")
    num1 = int(input("Enter First Number: "))
    num2 = int(input("Enter Second Number: "))
    try:
        result = num1 / num2
        print("Result =", result)
    except ZeroDivisionError:
        print("Cannot divide by zero")
    numbers = [10, 20, 30]
    print(numbers[5])
except ValueError:
    print("Invalid Input")
except IndexError:
    print("List index out of range")
```

4. Write a program to execute a statement using Finally Block.

```python
try:
    num1=int(input("Enter First Number: "))
    num2=int(input("Enter Second Number: "))
    result=num1/num2
    print("result=",result)
except ZeroDivisionError as es:
    print(es)
finally:
    print("program execution completed")
```

5. Write a program to handle exceptions with raise Keyword.

```python
try:
    num1=int(input('enter a numer'))
    if num1<18:
        raise ValueError ('age must be greater than 18')
    else:
        print(num1)
except ValueError as e1:
    print(e1)
```
