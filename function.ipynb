1. Write a Python function that accepts a string and counts the number of upper
and lowercase letters.
●Sample String: 'The quick Brown Fox'
●Expected Output: No. of uppercase characters: 3
●No. of Lower-case Characters: 12

```python
def func(str):
    upper_count=0
    lower_count=0
    for ch in str:
        if ch.isupper():
            upper_count+=1
        else:
            lower_count+=1
    print('no.of upper case:',upper_count)
    print('No. of Lower-case Characters:',lower_count)
func('sanDeeP')

```

2. Write a Python function that takes a list and returns a new list with distinct
elements from the first list,
●Sample List: [1,2,3,3,3,3,4,5]
●Unique List: [1, 2, 3, 4, 5]

```python
def func(list):
    unique=[]
    for i in list:
        if i not in unique:
            unique.append(i)
    print(unique)
func([1,2,3,3,3,3,4,5])
```

3. Write a Python function that checks whether a passed string is a palindrome
or not.


```python
def func(str):
    rev=''
    for i in range(len(str)-1,-1,-1):
        rev=rev+str[i]
    if str==rev:
        print(f'{str} is palindrome')
    else:
        print(f'{str} is not palindrome')
func('madam')
```

●Original list: [12, 0, None, 23, None,-55, 234, 89, None, 0, 6,-12]

8. Write a Python function to check whether a number is "Perfect" or not. In
number theory, a perfect number is a positive integer that is equal to the sum
of its proper positive divisors, that is, the sum of its positive divisors excluding
the number itself (also known as its aliquot sum). Equivalently, a perfect
number is a number that is half the sum of all of its positive divisors (including
itself)


```python
def func(number):
    total=0
    for i in range(1,number):
        if number%i==0:
            total+=i
    if total==number:
        print('it is a perfect number')
    else:
        print('not a perfect number')
func(int(input('enter a number')))
```

10. Write a Python program to create a function that takes one argument, and
that argument will be multiplied with an unknown given number, Sample
Output,
•Double the number of 15 = 30
•Triple the number of 15 = 45
•Quadruple the number of 15 = 60
•Quintuple the number 15 = 75

```python
def func(num):
    return lambda x:x*num
double=func(2)
triple=func(3)
quadruple = func(4)
quintuple = func(5)
print('double',double(2))
print('double',triple(2))
print('double',quadruple(2))
print('double',quintuple(2))
```

12. Write a Python program that creates a list of tuples, each containing a city
name and its population. Use the filter function to extract cities with a population
over 2 million.

```python
cities = [
    ("Hyderabad", 7000000),
    ("Vizag", 1800000),
    ("Mumbai", 1200000),
    ("Chennai", 5000000),
    ("Warangal", 900000),
    ('tokyo',3000000)
]
result=list(filter(lambda x:x[1] > 2000000,cities))
print(f'cities with population over 2million: {result}')
```
