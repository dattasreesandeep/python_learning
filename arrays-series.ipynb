1. Program to copy all elements of one array into another array.


```python
#array based
import array
arr1=array.array('i',(1,2,3,4))
arr2=array.array('i',[])
for i in arr1:
    arr2.append(i)
print('original array',arr1)
print('copied array',arr2)
```

```python
#list based
arr1 = [10, 20, 30, 40, 50]
arr2 = []
for element in arr1:
    arr2.append(element)
print("Original Array:", arr1)
print("Copied Array:", arr2)
```

2. Program to find the frequency of each element in the array.


```python
#array based
arr1=array.array('i',(1,2,3,4,1,3))
frequency={}
for i in arr1:
    if i in frequency:
        frequency[i]+=1
    else:
        frequency[i]=1
print(frequency)
```

```python
#list based

arr = [1, 2, 2, 3, 1, 4, 2]

freq = {}

# count frequency
for element in arr:

    if element in freq:
        freq[element] += 1
    else:
        freq[element] = 1

print(freq)
```

3. Program to print the duplicate elements of an array.


```python
#arraybased
import array
arr=array.array('i',[1,2,3,2, 1,5])
duplicates = []
for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] == arr[j] and arr[i] not in duplicates:
            duplicates.append(arr[i])
print("Duplicate Elements:", duplicates)
```

4. Program to print the elements of an array in reverse order without sorting.


```python
import array
arr = array.array('i', [10, 20, 30, 40, 50])
for i in range(len(arr) - 1, -1, -1):
    print(arr[i], end=' ')
```

5. Program to find Largest Number and smallest in an Array.


```python
import array
arr = array.array('i', [10, 5, 80, 2, 25])
largest = arr[0]
smallest = arr[0]
for element in arr:
    if element > largest:
        largest = element
    if element < smallest:
        smallest = element
print("Largest =", largest)
print("Smallest =", smallest)
```

6. Program to sort the elements of an array in descending order.


```python
import array
arr = array.array('i', [5, 2, 8, 1, 9])
for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] < arr[j]:
            arr[i], arr[j] = arr[j], arr[i]
print(arr)
```

7. Program to print the sum of all the items of the array.

 8. Program to print Odd and Even Numbers from an Array.

```python
import array
arr = array.array('i', [10, 15, 22, 33, 40, 55])
print("Even Numbers:")
for element in arr:
    if element % 2 == 0:
        print(element, end=' ')
print("\nOdd Numbers:")
for element in arr:
    if element % 2 != 0:
        print(element, end=' ')
```

9. Program to check whether a given number is a happy number or not?


```python
num = int(input("Enter Number: "))
original = num
visited = []
while num != 1 and num not in visited:
    visited.append(num)
    total = 0
    while num > 0:
        digit = num % 10
        total += digit * digit
        num //= 10
    num = total
if num == 1:
    print(original, "is Happy Number")
else:
    print(original, "is Not Happy Number")
```

10. Program to Find the Sum of the Series 1/1+1/2+1/3+…1/N.

```python
n = int(input("Enter N:"))
total = 0
for i in range(1, n + 1):
    total += 1 / i
print("Sum =", total)
```

11. Program to Find the Sum of the Series 1/1+1/2^2+1/3^2+1/4^2+1/5^2.

```python
n = int(input("Enter N: "))
total = 0
for i in range(1, n + 1):
    total += 1 / (i * i)
print("Sum =", total)
```

12. Program to Find the Sum of the Series 1/1^0+1/2^1+1/3^2+1/4^3+1/5^4.

14. Program to Find the Sum of the Series 1/1!+1/2!+1/3!+…1/N!

```python
n = int(input("Enter N: "))
total = 0
factorial = 1
for i in range(1, n + 1):
    factorial *= i
    total += 1 / factorial
print("Sum =", total)
```
