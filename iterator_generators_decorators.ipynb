**Decorators**

```python
def decorator_function(original_function):
    def wrapper():
        print("Before Function")
        original_function()
        print("After Function")
    return wrapper
def display():
    print("Hello")
decorated = decorator_function(display)
decorated()
```

```python
def decorator_function(original_function):
    def wrapper():
        print("Before Function")
        original_function()
        print("After Function")
    return wrapper
@decorator_function
def display():
    print("Hello")
display()
```
