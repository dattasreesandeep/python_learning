```python
n = 5

for i in range(1, n + 1):

    # spaces
    print(' ' * (n - i), end='')


    # ascending numbers
    for j in range(1, i + 1):

        print(j, end='')


    # descending numbers
    for j in range(i - 1, 0, -1):

        print(j, end='')


    print()


for i in range(n - 1, 0, -1):

    # spaces
    print(' ' * (n - i), end='')


    # ascending numbers
    for j in range(1, i + 1):

        print(j, end='')


    # descending numbers
    for j in range(i - 1, 0, -1):

        print(j, end='')


    print()
```
