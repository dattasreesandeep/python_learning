3. Program to find maximum and minimum occurring character in a string

```python
s1='sandeep'
max_ch=s1[0]
min_ch=s1[0]
for ch in s1:
    if s1.count(ch) > s1.count(max_ch):
        max_ch=ch
    if s1.count(ch) < s1.count(min_ch):
        min_ch=ch
print(f'{max_ch} is max occuring')
print(f'{min_ch} is min occuring')

```

4. Program to count the total number of Words,Numbers,Spl characters,vowels
and consonants in a string

```python
s3 ='hi my name is #@123@3Aue'
words=0
numbers=0
spl=0
vowels=0
consonants=0
v='aeiouAEIOU'
for ch in s3:
    if ch.isalpha():
        if ch in v:
            vowels+= 1
        else:
            consonants+=1
    elif ch.isdigit():
        numbers+=1
    elif ch !=' ':
        spl+=1
words = len(s3.split())
print("words =",words)
print("numbers=",numbers)
print("special characters =",spl)
print("vowels=", vowels)
print("consonants=",consonants)
```

14.Program to replace each word’s first letter with its previous character in a
given sentence.

```python
s=str(input("sentence"))
words =s.split()
for i in range(len(words)):
    first = words[i][0]
    prev = chr(ord(first) - 1)
    words[i] = prev + words[i][1:]
result = ' '.join(words)
print(result)
```

17. Write a program to find the percentage of uppercase, lowercase, vowels,
consonants digits, and special characters in a String.

```python
s ='$$~#Sandeep is learning ~₹$$₹Python.'
u=l=v=c=d=sp=0
for ch in s:
    if ch.isupper():
        u+= 1
    if ch.islower():
        l+= 1
    if ch.lower() in "aeiou" and ch.isalpha():
        v+= 1
    elif ch not in 'aeiou' and ch.isalpha():
        c+= 1
    elif ch.isdigit():
        d+= 1
    else:
        sp+= 1
total = len(s)
print("Uppercase Percentage:",u * 100/total)
print("Lowercase Percentage:", l * 100 / total)
print("Vowels Percentage:", v * 100 / total)
print("Consonants Percentage:", c * 100 / total)
print("Digits Percentage:", d * 100 / total)
print("Special Characters Percentage:", sp * 100 / total)
print(total)
```

18. Write a program to Print strings in reverse order word by word.
